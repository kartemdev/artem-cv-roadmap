```python
from pathlib import Path

import matplotlib.patches as patches
import matplotlib.pyplot as plt
from PIL import Image

from cv_lab.config import DATA_DIR

raw_data_dir = Path("../../") / DATA_DIR / "raw"

yamaha_path = raw_data_dir / "yamaha_mt_09.jpg"
yamaha_image = Image.open(yamaha_path)

true_box = [50, 50, 200, 300]
pred_box = [45, 60, 210, 290]
confidence = 0.94

fig, ax = plt.subplots(1, 2, figsize=(12, 6))

ax[0].imshow(yamaha_image)
ax[0].axis("off")
ax[0].set_title("Правильный ответ (Ground Truth)", fontsize=14, color="darkgreen")

rect_true = patches.Rectangle(
    (true_box[0], true_box[1]),
    true_box[2],
    true_box[3],
    linewidth=3,
    edgecolor="green",
    facecolor="none",
)

ax[0].add_patch(rect_true)

ax[1].imshow(yamaha_image)
ax[1].axis("off")
ax[1].set_title(
    f"Предсказание модели (Conf: {confidence:.2f})", fontsize=14, color="darkred"
)

rect_pred = patches.Rectangle(
    (pred_box[0], pred_box[1]),
    pred_box[2],
    pred_box[3],
    linewidth=3,
    edgecolor="red",
    facecolor="none",
)

ax[1].add_patch(rect_pred)

plt.tight_layout()
plt.show()
```
