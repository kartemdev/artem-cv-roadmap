```python
import random

import pandas as pd

moto_base = {
    "Yamaha": [
        ("MT-09", 2014, 2026, 115, 119),
        ("MT-09 SP", 2018, 2026, 115, 119),
        ("MT-07", 2014, 2026, 73, 75),
        ("MT-03", 2016, 2026, 42, 42),
        ("YZF-R1", 1998, 2025, 150, 200),
        ("YZF-R6", 1999, 2020, 116, 120),
        ("Ténéré 700", 2019, 2026, 73, 73),
        ("FJR1300", 2001, 2024, 143, 145),
        ("XSR900", 2016, 2026, 115, 119),
    ],
    "Honda": [
        ("CB750 Hornet", 2023, 2026, 91.8, 91.8),
        ("CB650R", 2019, 2026, 94, 95),
        ("CBR1000RR Fireblade", 2004, 2026, 172, 217),
        ("CRF1100L Africa Twin", 2020, 2026, 102, 102),
        ("NC750X", 2014, 2026, 54, 58),
        ("CB500X", 2013, 2024, 47, 47),
        ("GL1800 Gold Wing", 2001, 2026, 118, 126),
        ("CBR650R", 2019, 2026, 94, 95),
    ],
    "Kawasaki": [
        ("Z900", 2017, 2026, 125, 125),
        ("Z650", 2017, 2026, 68, 68),
        ("Ninja H2", 2015, 2026, 200, 231),
        ("Ninja 400", 2018, 2024, 45, 49),
        ("Ninja ZX-10R", 2004, 2026, 175, 203),
        ("Versys 650", 2007, 2026, 64, 69),
        ("Vulcan S", 2015, 2026, 61, 61),
    ],
    "Suzuki": [
        ("GSX1300R Hayabusa", 1999, 2026, 175, 197),
        ("GSX-R1000", 2001, 2024, 160, 202),
        ("SV650", 1999, 2026, 64, 76),
        ("V-Strom 650", 2004, 2026, 64, 71),
        ("GSX-S750", 2015, 2022, 114, 114),
        ("Katana", 2019, 2026, 150, 152),
    ],
    "BMW": [
        ("R1250GS", 2019, 2024, 136, 136),
        ("R1300GS", 2024, 2026, 145, 145),
        ("S1000RR", 2009, 2026, 193, 210),
        ("F850GS", 2018, 2024, 95, 95),
        ("G310GS", 2017, 2026, 34, 34),
        ("R nineT", 2014, 2024, 110, 110),
    ],
    "Ducati": [
        ("Monster 821", 2014, 2020, 109, 112),
        ("Panigale V4", 2018, 2026, 211, 216),
        ("Multistrada V4", 2021, 2026, 170, 170),
        ("Scrambler Icon", 2015, 2026, 73, 75),
        ("Streetfighter V4", 2020, 2026, 208, 208),
    ],
    "KTM": [
        ("390 Duke", 2013, 2026, 43, 44),
        ("890 Adventure", 2021, 2024, 105, 105),
        ("1290 Super Duke R", 2014, 2023, 173, 180),
        ("690 SMC R", 2008, 2026, 62, 74),
    ],
    "Triumph": [
        ("Street Triple 765 R", 2017, 2026, 118, 120),
        ("Street Triple 765 RS", 2017, 2026, 123, 130),
        ("Speed Triple 1200 RS", 2021, 2026, 180, 183),
        ("Speed Triple 1200 RR", 2021, 2026, 180, 183),
        ("Tiger 900", 2020, 2026, 95, 108),
        ("Bonneville T120", 2016, 2026, 80, 80),
    ],
    "Harley-Davidson": [
        ("Iron 883", 2009, 2022, 51, 51),
        ("Pan America 1250", 2021, 2026, 152, 152),
        ("Fat Boy", 1990, 2026, 67, 94),
    ],
}

makers, models, years, powers = [], [], [], []

for maker, model_list in moto_base.items():
    for model_info in model_list:
        model_name, start_yr, end_yr, min_hp, max_hp = model_info

        available_years = list(range(start_yr, end_yr + 1))
        sample_years = random.sample(available_years, min(4, len(available_years)))

        for yr in sample_years:
            if start_yr == end_yr:
                hp = min_hp
            else:
                progress = (yr - start_yr) / (end_yr - start_yr)
                hp = min_hp + (max_hp - min_hp) * progress

            makers.append(maker)
            models.append(model_name)
            years.append(str(yr))
            powers.append(str(round(hp, 1) if hp % 1 != 0 else int(hp)))

moto_dict = {
    "Maker": makers,
    "Model": models,
    "Year of made": years,
    "Power hp": powers,
}

df = pd.DataFrame(moto_dict)

print(df)
```
