```python
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image

from cv_lab.config import PROJECT_ROOT
from cv_lab.io import collect_image_paths

image_paths = collect_image_paths(PROJECT_ROOT)
first_image = Image.open(image_paths[0])

plt.imshow(first_image)
```

```python
img_array = np.array(Image.open(image_paths[0]))

print(img_array.shape)
```

```python
print(img_array.dtype)
```

```python
print(img_array.min())
```

```python
print(img_array.max())
```

```python
print(img_array.mean())
```

```python
img_array = img_array.astype(np.float32)

print(img_array.dtype)
print(img_array[0][0])

img_array = img_array / 255.0

print(img_array[0][0])
```

```python
img_array = np.clip(img_array, 0, 255)

print(img_array)
```

```python
tensor_image = img_array.transpose(2, 0, 1)

print(tensor_image.shape)

plt_format = tensor_image.transpose(1, 2, 0)

plt_format = plt_format * 255

plt.imshow(plt_format.astype(np.uint8))
plt.show()
```
