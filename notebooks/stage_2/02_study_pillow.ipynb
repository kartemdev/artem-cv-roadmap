```python
import numpy as np
from PIL import Image

from cv_lab.config import DATA_DIR
from cv_lab.io import collect_image_paths

image_paths = collect_image_paths(DATA_DIR)
img_array = np.array(Image.open(image_paths[5]).convert("L"))

image = Image.fromarray(img_array)

image
```

```python
shape = img_array.shape
aspect_ratio = shape[0] / shape[1]

new_width = 770
new_height = int(new_width * aspect_ratio)

image.resize((new_width, new_height))
```
