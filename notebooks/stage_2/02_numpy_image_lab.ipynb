```python
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image

from cv_lab.config import RAW_DATA_DIR
from cv_lab.io import collect_image_paths

image_paths = collect_image_paths(RAW_DATA_DIR)
image = Image.open(image_paths[11])

arr = np.array(image)
```

```python
print(arr.shape, arr.dtype, arr.min(), arr.max())
```

```python
plt.imshow(arr)
plt.axis("off")
```

```python
red = arr[:, :, 0]
green = arr[:, :, 1]
blue = arr[:, :, 2]

print(red.shape, green.shape, blue.shape)
```

```python
gray = 0.2126 * red + 0.7152 * green + 0.0722 * blue

print(gray.shape, gray.dtype, gray.min(), gray.max())
```

```python
plt.imshow(gray, cmap="gray")
plt.axis("off")
```

```python
normalized = arr.astype(np.float32) / 255.0

print(normalized.dtype, normalized.min(), normalized.max())
```

```python
crop = arr[100:300, 200:500, :]

crop.shape
```

```python
plt.imshow(crop)
plt.axis("off")
```

```python
mask = gray > gray.mean()

print(mask.shape, mask.dtype)
```

```python
plt.imshow(mask, cmap="gray")
plt.axis("off")
```
